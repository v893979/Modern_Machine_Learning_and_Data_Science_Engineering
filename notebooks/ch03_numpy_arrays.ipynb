## 📌 Notebook Overview
Learn matrix operations, feature standardization (Z-score), broadcasting, and linear equation evaluation ($z = \mathbf{X}\mathbf{w} + b$).

### 1. Feature Standardization (Z-Score Normalization)

Transform multi-dimensional feature scales so every feature has $\mu=0$ and $\sigma=1$:

```python
import numpy as np

# 5 samples x 3 features (Age, Income, Credit Score)
X = np.array([
    [25, 45000, 680],
    [42, 85000, 740],
    [31, 52000, 620],
    [55, 120000, 790],
    [22, 38000, 590]
], dtype=float)

mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
X_scaled = (X - mean) / std

print("Feature Means:\n", np.round(mean, 2))
print("Feature Stds:\n", np.round(std, 2))
print("Standardized Matrix:\n", np.round(X_scaled, 3))
```

### 2. Linear Algebra: Matrix Dot Product

Compute model activation scores $z = \mathbf{X}\mathbf{w} + b$ across all samples in one vectorized operation:

```python
weights = np.array([0.2, 0.5, 0.8])
bias = -0.1
z_scores = X_scaled @ weights + bias
print("Computed Scores z = X @ w + b:\n", np.round(z_scores, 3))
```
