## 📌 Notebook Overview
Understand how continuous word vectors solve sparse bag-of-words limitations and compute geometric vector analogies: $\mathbf{v}_{\text{King}} - \mathbf{v}_{\text{Man}} + \mathbf{v}_{\text{Woman}} = \mathbf{v}_{\text{Queen}}$.

### 1. Defining Dense Semantic Vectors

Simulate multi-dimensional word embeddings across Royalty, Gender, and Tech dimensions:

```python
import numpy as np

def cosine_similarity(u, v):
    dot = np.dot(u, v)
    norm = np.linalg.norm(u) * np.linalg.norm(v)
    return dot / norm if norm > 0 else 0.0

vocab = {
    "king":    np.array([0.95,  0.88, -0.10, -0.05, 0.60]),
    "queen":   np.array([0.96, -0.12,  0.90, -0.05, 0.55]),
    "man":     np.array([0.05,  0.92, -0.15,  0.02, 0.50]),
    "woman":   np.array([0.05, -0.15,  0.94,  0.02, 0.48]),
    "prince":  np.array([0.88,  0.85, -0.08, -0.02, 0.20]),
    "princess":np.array([0.89, -0.08,  0.88, -0.02, 0.18]),
    "apple":   np.array([0.02,  0.01,  0.01,  0.85, 0.10]),
    "computer":np.array([0.01,  0.02,  0.00,  0.95, 0.12]),
}

print("Cosine Similarity(king, queen):", round(cosine_similarity(vocab["king"], vocab["queen"]), 4))
print("Cosine Similarity(king, man):  ", round(cosine_similarity(vocab["king"], vocab["man"]), 4))
```

### 2. Solving the Classic Vector Analogy

Compute $\text{Target} = \mathbf{v}_{\text{King}} - \mathbf{v}_{\text{Man}} + \mathbf{v}_{\text{Woman}}$ and rank candidates:

```python
target = vocab["king"] - vocab["man"] + vocab["woman"]

scores = {}
for word, vec in vocab.items():
    if word not in ["king", "man", "woman"]:
        scores[word] = cosine_similarity(target, vec)

for rank, (word, sim) in enumerate(sorted(scores.items(), key=lambda x: x[1], reverse=True)[:3], 1):
    print(f"Rank {rank}: {word:<10} | Cosine Similarity = {sim:.4f}")
```
