## 1. Variables, Data Types & Dynamic Typing
In Python, a **variable** is simply a labeled storage box in computer memory. You assign data to a variable using the single equals sign (`=`).

Python features **dynamic typing**—you don't have to declare the type ahead of time; Python figures it out automatically.


```python
# Real-World Healthcare Scenario: Patient Record
patient_id = 'P-10492'           # String (str) - textual characters
patient_age = 42                 # Integer (int) - whole number
body_temperature = 98.6          # Float (float) - decimal number
has_preexisting_condition = True # Boolean (bool) - True or False

print('Patient ID:   ', patient_id, ' | Type:', type(patient_id).__name__)
print('Age:          ', patient_age, '     | Type:', type(patient_age).__name__)
print('Temperature:  ', body_temperature, '   | Type:', type(body_temperature).__name__)
print('Pre-existing: ', has_preexisting_condition, '   | Type:', type(has_preexisting_condition).__name__)

# Multiple assignment in one line
systolic, diastolic = 120, 80
print(f'Blood Pressure: {systolic}/{diastolic} mmHg')
```

## 2. Type Casting & Mathematical Formulas (BMI Calculation)
When data arrives from CSV files or web APIs, numbers often enter Python as text strings (e.g., `'78.5'`). Trying to do math on a string triggers a `TypeError`. We must **cast** (convert) the string into a numeric type.


```python
# Raw string inputs from digital medical sensor
raw_weight = '78.5'
raw_height = '1.75'

# Type casting from str to float
weight_kg = float(raw_weight)
height_m = float(raw_height)

# Calculate Body Mass Index: weight / (height ** 2)
bmi = weight_kg / (height_m ** 2)
print(f'Patient Weight: {weight_kg} kg | Height: {height_m} m')
print(f'Calculated BMI: {bmi:.2f}')
```

## 3. Arithmetic, Comparison & Logical Operators (Food Delivery Fee Engine)
Python supports standard math (`+`, `-`, `*`, `/`) along with integer floor division (`//`), modulo remainder (`%`), and exponentiation (`**`).


```python
base_delivery_fee = 3.99
trip_distance_miles = 6.4
per_mile_rate = 1.25
is_rush_hour = True
is_bad_weather = False
customer_is_vip = True

# Calculate distance fee
distance_fee = trip_distance_miles * per_mile_rate
subtotal = base_delivery_fee + distance_fee

# Logical operators: and, or, not
surge_multiplier = 1.5 if (is_rush_hour or is_bad_weather) else 1.0
discount = 2.00 if (customer_is_vip and surge_multiplier <= 1.5) else 0.00

total_fee = (subtotal * surge_multiplier) - discount
print(f'Subtotal:     ${subtotal:.2f}')
print(f'Surge Factor: {surge_multiplier}x')
print(f'VIP Discount:-${discount:.2f}')
print(f'Final Fee:    ${total_fee:.2f}')
```

## 4. Decision Making with Conditionals (`if`, `elif`, `else`)
Conditionals allow software to make decisions based on defined thresholds.


```python
def triage_patient(heart_rate: int, spo2: float) -> str:
    if spo2 < 90.0 or heart_rate > 130:
        return 'RED: Emergency Resuscitation Required'
    elif (90.0 <= spo2 < 94.0) or (110 < heart_rate <= 130):
        return 'YELLOW: Urgent Medical Evaluation'
    elif spo2 >= 94.0 and (60 <= heart_rate <= 100):
        return 'GREEN: Normal Stable Vital Signs'
    else:
        return 'BLUE: Ambulatory / Low Priority'

print(triage_patient(heart_rate=145, spo2=88.5))
print(triage_patient(heart_rate=72,  spo2=98.0))
```

## 5. Loops: Repeating Work with `while` and `for`
- Use a **`while` loop** when you need to repeat an action until a condition becomes false (e.g. charging an EV battery).
- Use a **`for` loop** when you need to iterate through a sequence (e.g. auditing transactions).


```python
# While loop: Simulating EV Battery Charging up to 80%
battery = 20.0
minute = 0
while battery < 80.0:
    minute += 5
    charge_added = 12.0 if battery < 50.0 else 6.0
    battery += charge_added
    print(f'Minute {minute:02d}: +{charge_added:.1f}% -> Current Battery: {min(battery, 100.0):.1f}%')
    if minute >= 40:  # Safety cutoff
        break

print(f'Charging Session Concluded at {battery:.1f}% in {minute} mins.')
```

```python
# For loop with zip() and enumerate()
regions = ['Downtown', 'Suburbs', 'Airport', 'Tech Corridor']
sales = [12500.0, 8900.0, 15400.0, 11800.0]
targets = [10000.0, 9500.0, 14000.0, 11000.0]

print('Regional Performance Summary:')
for region, actual, target in zip(regions, sales, targets):
    status = 'EXCEEDED' if actual >= target else 'MISSED'
    print(f'  {region:<15} Actual: ${actual:>9,.2f} | Target: ${target:>9,.2f} -> [{status}]')
```

## 6. Reusable Functions (FinTech Debt-to-Income Evaluator)
Functions group reusable logic into organized, testable building blocks.


```python
def evaluate_applicant(name: str, monthly_income: float, monthly_debt: float, credit_score: int) -> tuple:
    '''Calculates Debt-to-Income (DTI) and returns an approval decision.'''
    dti = monthly_debt / monthly_income
    if credit_score < 620:
        return False, dti, 'SUBPRIME_CREDIT'
    elif dti > 0.43:
        return False, dti, 'EXCESSIVE_DTI'
    else:
        return True, dti, 'APPROVED'

# Testing applicants
for name, inc, debt, score in [('Sarah', 7500, 1800, 740), ('Marcus', 5000, 2600, 680)]:
    approved, dti, reason = evaluate_applicant(name, inc, debt, score)
    print(f'{name}: Decision={approved} | DTI={dti*100:.1f}% | Reason={reason}')
```
