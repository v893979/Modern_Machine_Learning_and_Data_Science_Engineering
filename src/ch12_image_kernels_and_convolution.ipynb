Implement 2D convolution and max pooling from scratch in pure NumPy, isolating edge boundaries with vertical Sobel filters.

### 1. 2D Spatial Convolution Engine

Slide a $3 \times 3$ kernel matrix over a pixel grid and compute dot products:

```python
import numpy as np

def convolve2d(image, kernel):
    i_h, i_w = image.shape
    k_h, k_w = kernel.shape
    out_h = i_h - k_h + 1
    out_w = i_w - k_w + 1
    output = np.zeros((out_h, out_w))
    for i in range(out_h):
        for j in range(out_w):
            patch = image[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(patch * kernel)
    return output

image = np.array([
    [10, 10, 10, 90, 90, 90],
    [10, 10, 10, 90, 90, 90],
    [10, 10, 10, 90, 90, 90],
    [10, 10, 10, 90, 90, 90],
    [10, 10, 10, 90, 90, 90],
    [10, 10, 10, 90, 90, 90],
], dtype=float)

sobel_vertical = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=float)

edge_map = convolve2d(image, sobel_vertical)
print("Sobel Edge Feature Map (4x4):\n", edge_map.astype(int))
```
