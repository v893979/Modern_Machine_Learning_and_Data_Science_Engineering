## 📌 Notebook Overview
Construct and train a modular Multi-Layer Perceptron (MLP) binary classifier in PyTorch with BCEWithLogitsLoss and AdamW.

### 1. PyTorch Model Architecture

Define a deep neural network with Linear layers, ReLU activations, and Dropout regularization:

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

class BinaryClassifierMLP(nn.Module):
    def __init__(self, in_features):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_features, 32),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 1)
        )
        
    def forward(self, x):
        return self.net(x)

model = BinaryClassifierMLP(in_features=8)
print(model)
```

### 2. Training Loop & Forward/Backward Propagation

Execute 5 epochs of gradient descent with the AdamW optimizer:

```python
torch.manual_seed(42)
X = torch.randn(1000, 8)
y = torch.randint(0, 2, (1000, 1)).float()

dataset = TensorDataset(X, y)
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

criterion = nn.BCEWithLogitsLoss()
optimizer = optim.AdamW(model.parameters(), lr=0.005, weight_decay=1e-4)

model.train()
for epoch in range(5):
    total_loss = 0.0
    for batch_X, batch_y in dataloader:
        optimizer.zero_grad()
        logits = model(batch_X)
        loss = criterion(logits, batch_y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * batch_X.size(0)
    print(f"Epoch {epoch+1:02d} / 05 | Average Loss: {total_loss / 1000:.4f}")
```
