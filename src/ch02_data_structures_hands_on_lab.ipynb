## 1. Lists: Slicing, Mutability & Essential Methods
Lists are the most versatile data structure in Python. They are **ordered** and **mutable** (can be modified in place).


```python
passengers = ['Sarah Connor', 'John Doe', 'Alice Smith', 'Bob Vance', 'Elena Rostova']
print(f'Original Manifest: {passengers}')

# Zero-based indexing and negative indexing
print('First (0):', passengers[0])
print('Last (-1):', passengers[-1])
print('Penultimate (-2):', passengers[-2])

# Slicing: [start:stop:step]
print('First 3 (:3):     ', passengers[:3])
print('Middle [1:3]:     ', passengers[1:3])
print('Alternate (::2):  ', passengers[::2])
print('Reversed (::-1):  ', passengers[::-1])

# Essential list mutations
passengers.append('Marcus Aurelius')
passengers.insert(0, 'VIP: Captain Smith')
print('Updated Manifest:', passengers)
```

## 2. Tuples: Immutability & Coordinate Protection
Tuples are defined with parentheses `()`. Unlike lists, tuples are **immutable**—once created, their elements cannot be changed, added, or deleted.


```python
warehouse_gps = (37.7749, -122.4194, 'San Francisco Distribution Hub')

# Tuple unpacking into distinct variables
lat, lon, label = warehouse_gps
print(f'Latitude: {lat} | Longitude: {lon} | Location: {label}')

# Demonstrating immutability protection
try:
    warehouse_gps[0] = 38.0000
except TypeError as e:
    print(f'Protected from accidental corruption! Error: {e}')
```

## 3. Sets: Fast Deduplication & Mathematical Operations (CyberShield Firewall)
Sets are defined with curly braces `{}` or `set()`. They store **unique** values only and enable fast $O(1)$ constant-time membership lookups.


```python
raw_server_ips = ['192.168.1.10', '10.0.0.5', '192.168.1.10', '172.16.0.4', '10.0.0.5', '192.168.1.99']
unique_ips = set(raw_server_ips)
print(f'Raw Count: {len(raw_server_ips)} -> Deduplicated Count: {len(unique_ips)}')

known_threat_ips = {'192.168.1.99', '203.0.113.5', '198.51.100.22', '172.16.0.4'}

# Set Operations:
active_threats = unique_ips & known_threat_ips  # Intersection
clean_traffic  = unique_ips - known_threat_ips  # Difference
all_monitored  = unique_ips | known_threat_ips  # Union

print('[BLOCKED] Active Threats (&):', active_threats)
print('[ALLOWED] Clean Traffic  (-):', clean_traffic)
```

## 4. Dictionaries: Modeling Real-World Entities
Dictionaries map unique keys to values. Always use `.get()` to access keys defensively without risking a `KeyError` crash.


```python
customer = {
    'account_id': 'ACC-8831',
    'name': 'Evelyn Reed',
    'tier': 'Gold',
    'spend': 14250.75,
    'metadata': {'industry': 'Healthcare', 'risk_score': 0.12}
}

# Safe lookup with default fallback
referral = customer.get('referral_code', 'NO_REFERRAL')
cust_name = customer['name']
cust_tier = customer['tier']
print(f'Customer: {cust_name} | Tier: {cust_tier} | Referral: {referral}')
```

## 5. End-to-End FinTech Case Study: 'ApexPay' Transaction Risk Engine
Let's see how a data architect designs a transaction auditing pipeline:  
- Moving from fragile parallel lists to unified dictionaries  
- Applying automated fraud screening filters  
- Exporting the audited records to a clean **Pandas DataFrame**


```python
import pandas as pd

# Unified database of transactions
transactions_db = [
    {'tx_id': 'TX-101', 'merchant': 'Metro Grocers', 'amount': 150.00,  'country': 'US', 'risk_score': 0.05, 'status': 'APPROVED'},
    {'tx_id': 'TX-102', 'merchant': 'Global Crypto',  'amount': 4800.00, 'country': 'RU', 'risk_score': 0.65, 'status': 'REVIEW'},
    {'tx_id': 'TX-103', 'merchant': 'Coffee Haven',   'amount': 32.50,   'country': 'US', 'risk_score': 0.01, 'status': 'APPROVED'},
    {'tx_id': 'TX-104', 'merchant': 'Wire Transfer',  'amount': 9200.00, 'country': 'NG', 'risk_score': 0.88, 'status': 'REVIEW'},
    {'tx_id': 'TX-105', 'merchant': 'London Books',   'amount': 210.00,  'country': 'GB', 'risk_score': 0.10, 'status': 'APPROVED'},
]

# Automated risk audit
high_risk_countries = {'RU', 'NG', 'KP', 'IR'}
for tx in transactions_db:
    if (tx['amount'] > 3000.0) or (tx['country'] in high_risk_countries) or (tx['risk_score'] > 0.70):
        tx['status'] = 'FLAGGED_FOR_AUDIT'

# Convert to Pandas DataFrame for executive reporting
df = pd.DataFrame(transactions_db)
print(df[['tx_id', 'merchant', 'amount', 'country', 'risk_score', 'status']])
total_vol = df['amount'].sum()
print(f'\nTotal Volume: ${total_vol:,.2f}')
```
