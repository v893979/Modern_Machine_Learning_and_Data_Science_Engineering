## 📌 Notebook Overview
Build an in-memory semantic vector search index, compute cosine similarities, and retrieve relevant domain documents for RAG context injection.

```python
# Environment & Data Auto-Download for Google Colab
import os
import urllib.request

os.makedirs('../data', exist_ok=True)
data_path = "../data/knowledge_base_rag.json"
if not os.path.exists(data_path):
    url = "https://raw.githubusercontent.com/v893979/Modern_Machine_Learning_and_Data_Science_Engineering/main/data/knowledge_base_rag.json"
    print(f"Downloading knowledge_base_rag.json from GitHub...")
    urllib.request.urlretrieve(url, data_path)
    print("Download complete!")
```

### 1. Ingestion & Embedding Generation

Index institutional knowledge manual documents:

```python
import json
import math

with open("../data/knowledge_base_rag.json", "r") as f:
    docs = json.load(f)

print(f"Loaded {len(docs)} documents into vector knowledge base.")
for d in docs:
    print(f"[{d['doc_id']}] {d['title']}")
```

### 2. Semantic Cosine Vector Retrieval

Query the vector store with user prompt: 'What is the maximum debt to income ratio allowed?'

```python
def mock_dense_embedding(text):
    keywords = ["mortgage", "credit", "fico", "dti", "income", "tax", "rate", "points", "underwriting", "down payment"]
    text_lower = text.lower()
    vec = [text_lower.count(kw) + 0.05 for kw in keywords]
    norm = math.sqrt(sum(x * x for x in vec))
    return [x / norm for x in vec]

query = "What is the maximum debt to income ratio allowed?"
query_vec = mock_dense_embedding(query)

scores = []
for doc in docs:
    d_vec = mock_dense_embedding(doc["title"] + " " + doc["content"])
    sim = sum(a * b for a, b in zip(query_vec, d_vec))
    scores.append((sim, doc))

scores.sort(key=lambda x: x[0], reverse=True)
top_doc = scores[0][1]
print(f"\nTop Match: [{top_doc['doc_id']}] {top_doc['title']}")
print(f"Content: {top_doc['content']}")
```
