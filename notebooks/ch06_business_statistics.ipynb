Perform rigorous statistical inference including two-sample independent t-tests for A/B testing and Chi-Square contingency tests.

### 1. Two-Sample Independent t-Test (A/B Testing)

Determine whether a new checkout design creates a statistically significant uplift in customer spend:

```python
import numpy as np
from scipy import stats

np.random.seed(42)
group_A = np.random.normal(loc=52.0, scale=12.0, size=150) # Control
group_B = np.random.normal(loc=56.5, scale=12.0, size=150) # Variation (New UI)

t_stat, p_val = stats.ttest_ind(group_A, group_B)
print(f"Group A Mean: ${np.mean(group_A):.2f}")
print(f"Group B Mean: ${np.mean(group_B):.2f}")
print(f"t-Statistic: {t_stat:.4f} | p-Value: {p_val:.4e}")

if p_val < 0.05:
    print("Conclusion: Reject H0! Statistically significant uplift detected (p < 0.05).")
else:
    print("Conclusion: Fail to reject H0.")
```

### 2. Chi-Square Test of Independence

Evaluate device type (Mobile vs Desktop) vs conversion behavior:

```python
contingency_table = np.array([
    [180, 820],  # Mobile: 180 Converted, 820 Bounced
    [240, 760]   # Desktop: 240 Converted, 760 Bounced
])
chi2, p_chi, dof, _ = stats.chi2_contingency(contingency_table)
print(f"Chi-Square: {chi2:.4f} | p-Value: {p_chi:.4e} | Degrees of Freedom: {dof}")
```
